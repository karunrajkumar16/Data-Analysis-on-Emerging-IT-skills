```python
from collections import Counter

# Convert all responses to lowercase, split by commas, and count skills
skills = df["Which IT skills are you currently learning or planning to learn?"].str.lower().str.split(",")

skill_list = [skill.strip() for sublist in skills for skill in sublist]  # flatten
skill_counts = Counter(skill_list)

# Show top 10 most common skills
skill_counts.most_common(10)
```

```python
# Clean messy characters and split properly
cleaned_skills = []

for response in df["Which IT skills are you currently learning or planning to learn?"].dropna():
    # lowercase
    response = response.lower()
    # replace ; and ) and ( with commas
    response = response.replace(";", ",").replace(")", "").replace("(", "")
    # split by comma
    parts = [p.strip() for p in response.split(",") if p.strip()]
    cleaned_skills.extend(parts)

# Count again
skill_counts_clean = Counter(cleaned_skills)

# Show top 10
skill_counts_clean.most_common(10)
```

```python
# Plot top 10 skills
import matplotlib.pyplot as plt

top_skills = dict(skill_counts_clean.most_common(10))

plt.figure(figsize=(10,6))
plt.bar(top_skills.keys(), top_skills.values(), color="skyblue", edgecolor="black")
plt.title("Top 10 IT Skills Students Are Learning", fontsize=14)
plt.xlabel("Skills", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
df["Which one IT skill do you believe is in the highest demand right now?"].value_counts()
```

```python
# Split multiple responses into individual skills
demand_skills = (
    df["Which one IT skill do you believe is in the highest demand right now?"]
    .dropna()  # remove missing values
    .str.lower()
    .str.split(";")  # split on semicolon
)

# Flatten into a single list
all_demand_skills = [skill.strip() for sublist in demand_skills for skill in sublist]

# Count frequencies
from collections import Counter
demand_counts = Counter(all_demand_skills)

# Convert to DataFrame for easy handling
demand_df = pd.DataFrame(demand_counts.most_common(10), columns=["Skill", "Count"])

# Show top 10
demand_df
```

```python
demand_df.plot(
    kind="bar",
    x="Skill",
    y="Count",
    legend=False,
    color="skyblue",
    edgecolor="black"
)

plt.title("Top 10 IT Skills in Highest Demand (Survey Results)")
plt.xlabel("Skill")
plt.ylabel("Number of Students")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Expand skills into multiple rows
skills_expanded = df.assign(
    Skills=df["Which IT skills are you currently learning or planning to learn?"].str.lower().str.split(";")
).explode("Skills")

# Clean whitespace
skills_expanded["Skills"] = skills_expanded["Skills"].str.strip()

# Take top 5 skills
top_skills = skills_expanded["Skills"].value_counts().head(5).index

# Take top 5 job roles
top_roles = df["Which IT job role are you most interested in?"].value_counts().head(5).index

# Filter for only top 5 skills and roles
filtered = skills_expanded[
    (skills_expanded["Skills"].isin(top_skills)) &
    (skills_expanded["Which IT job role are you most interested in?"].isin(top_roles))
]

# Count Skills vs Job Roles
skill_role_counts = filtered.groupby(
    ["Skills", "Which IT job role are you most interested in?"]
).size().unstack(fill_value=0)

# Plot grouped bar chart
skill_role_counts.T.plot(kind="bar", figsize=(10,6))

plt.title("Top 5 IT Skills vs Top 5 Job Roles")
plt.xlabel("Job Roles")
plt.ylabel("Number of Students")
plt.legend(title="Skills")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Example dataset (you can adjust with your actual survey data if structured)
data = {
    "Skill": ["AI/ML", "Data Science", "Cloud", "Web Dev", "Java", "CSS"],
    "3-5 LPA": [10, 5, 2, 12, 15, 14],
    "5-8 LPA": [15, 12, 6, 8, 10, 9],
    "8-12 LPA": [20, 15, 10, 5, 3, 2],
    "12+ LPA": [12, 8, 7, 1, 0, 0]
}

df_salary_skills = pd.DataFrame(data)

# Plot grouped bar chart
plt.figure(figsize=(10,6))
x = range(len(df_salary_skills))

bar_width = 0.2
plt.bar([i - 1.5*bar_width for i in x], df_salary_skills["3-5 LPA"], width=bar_width, label="3-5 LPA")
plt.bar([i - 0.5*bar_width for i in x], df_salary_skills["5-8 LPA"], width=bar_width, label="5-8 LPA")
plt.bar([i + 0.5*bar_width for i in x], df_salary_skills["8-12 LPA"], width=bar_width, label="8-12 LPA")
plt.bar([i + 1.5*bar_width for i in x], df_salary_skills["12+ LPA"], width=bar_width, label="12+ LPA")

plt.xticks(x, df_salary_skills["Skill"], rotation=45, ha="right")
plt.ylabel("Number of Students")
plt.title("Skills vs Salary Expectation")
plt.legend()
plt.tight_layout()
plt.show()
```
